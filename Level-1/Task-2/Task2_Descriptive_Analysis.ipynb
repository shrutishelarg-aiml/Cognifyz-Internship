IMPORTING LIBRARIES


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

LOADING DATASET

```python
df = pd.read_csv("/content/Dataset  (11).csv")
df
```

DATASET OVERVIEW

```python
print("Dataset shape: ",df.shape)
df.info()
```

STATISTICAL SUMMARY

```python
#Mean
df.mean(numeric_only=True)
```

```python
#Median
df.median(numeric_only=True)
```

```python
#Mode
df.mode(numeric_only=True)
```

```python
#Standard Deviation
df.std(numeric_only=True)
```

COUNTRY CODE DISTRIBUTION

```python
country_counts= df['Country Code'].value_counts()
country_counts
```

VISUALISATIONS

```python
plt.figure(figsize=(8,5))

country_counts.head(10).plot(kind='bar')

plt.title("Distribution of Country Codes")
plt.xlabel("Country Code")
plt.ylabel("Number of Restaurants")

plt.show()
```

CITY DISTRIBUTIONS

```python
city_counts = df['City'].value_counts()

print(city_counts.head(10))
```

VISUALISATIONS

```python
plt.figure(figsize=(10,5))

sns.barplot(
    x=city_counts.head(10).index,
    y=city_counts.head(10).values
)

plt.xticks(rotation=45)

plt.title("Top 10 Cities by Restaurant Count")

plt.show()
```

CUISINE DISTRIBUTION

```python
cuisine_counts = df['Cuisines'].value_counts()

print(cuisine_counts.head(10))
```

VISUALISATIONS

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=cuisine_counts.head(10).values,
    y=cuisine_counts.head(10).index
)

plt.title("Top 10 Cuisines")

plt.show()
```

FINDINGS

```python
print("Top City:")
print(df['City'].value_counts().idxmax())

print("\nTop Cuisine:")
print(df['Cuisines'].value_counts().idxmax())
```
